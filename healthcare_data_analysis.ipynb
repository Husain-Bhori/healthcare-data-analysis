Healthcare Data Analysis (Python)

This project demonstrates basic data analysis using Python on a simulated healthcare dataset.
The goal is to explore the relationship between hospital visit frequency and annual medical cost.

```python
#Import Libraries

import pandas as pd
import matplotlib.pyplot as plt
from IPython.display import display

#Create Healthcare Dataset
data = {"Age": [23, 45, 67, 34, 52, 60], "Annual_Visits": [1, 3, 5, 2, 4, 6], "Annual_Cost_USD": [200, 900, 1800, 400, 1200, 2000]}

df = pd.DataFrame(data)

display(df)
display(df.mean())

#Relationship Analysis
display(df[["Annual_Visits", "Annual_Cost_USD"]].corr())


#Creating Graph
plt.scatter(df["Annual_Visits"], df["Annual_Cost_USD"])
plt.xlabel("Annual Hospital Visits")
plt.ylabel("Annual Medical Cost (USD)")
plt.title("Relationship Between Hospital Visits and Cost")
plt.show()
```

Analysis Summary:

This project explores a simulated healthcare dataset using Python.
Summary statistics and correlation analysis suggest that higher hospital visit frequency is associated with increased annual medical costs.
A scatter plot was used to visualize this relationship.
